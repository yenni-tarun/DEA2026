```python
%python

df1 = spark.read \
    .option("header", True) \
    .option("inferSchema", True) \
    .csv("/Volumes/workspace/default/usage_metrics/mobile_os_usage.csv")

display(df1)
```

```python
%python

df1 = df1.withColumnRenamed(
    "Mobile Operating System",
    "Mobile_Operating_System"
).withColumnRenamed(
    "Percent of Usage",
    "Percent_of_Usage"
)

df1.write.mode("overwrite").saveAsTable("default.mobile_os_usage")
```

```python
%python

display(spark.sql("SELECT * FROM default.mobile_os_usage"))
```
